## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd
df = pd.read_csv("../../data/raw/content_refresh_anonymized.csv")
# Total inventory and active decay volume
total_pages = len(df)
declining_pages = (df["trend_direction"] == "down").sum()
declining_pct = (declining_pages / total_pages) * 100
# High-demand stale pages (Age >= 180 days AND Impressions >= 500)
stale_high_demand = df[(df["content_age_days"] >= 180) & (df["impressions_90d"] >= 500)]
stale_count = len(stale_high_demand)
# Page-one decay candidate pages (Position 1-10 AND Age >= 180 days)
page_one_decay = df[(df["avg_position"] > 0) & (df["avg_position"] <= 10) & (df["content_age_days"] >= 180)]
page_one_count = len(page_one_decay)
print(f"1. Total pages analyzed: {total_pages:,}")
print(f"   Pages in active decline: {declining_pages:,} ({declining_pct:.1f}% of" " total)")
print(f"2. High-demand stale pages (age >= 180d, impressions >= 500):" f" {stale_count:,}")
print(f"3. Page-one decay risks (position <= 10, age >= 180d):" f" {page_one_count:,}")
```
